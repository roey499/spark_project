```python
from pyspark.sql import SparkSession
spark =SparkSession.builder\
.master("local")\
.appName('ex2_airports')\
.getOrCreate()
```

#DOCKER COMPOSE UP - mariadb,hive_metastore,dev_env,minio

```python

# Utilize Spark's read.csv method to read the CSV files from the S3 location.
#specify that the CSV has a header, so the first row will be used as column names.
airports_raw_df = spark.read.csv('s3a://pyspark/data/raw/airports/', header=True)
```

```python
from pyspark.sql import functions as F
from pyspark.sql import types as T
# transform the data by selecting only the columns you need and casting the 'airport_id' column to IntegerType.
#alias it to keep the name consistent.
airports_df = airports_raw_df.select(
F.col('airport_id').cast(T.IntegerType()).alias('airport_id'),
F.col('city'),
F.col('state'),
F.col('name'))
```

```python
# write the transformed DataFrame back to S3 in Parquet format.
airports_df.write.parquet('s3a://pyspark/data/source/airports/', mode='overwrite')
```
